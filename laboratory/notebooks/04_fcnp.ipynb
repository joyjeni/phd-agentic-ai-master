# O4 FCNP — Feedback-conditioned context pruning


```python
import sys
from pathlib import Path
for candidate in [Path.cwd(), Path('/kaggle/working')]:
    if (candidate / 'python').exists():
        sys.path.insert(0, str(candidate / 'python'))
        break
from phd_agentic.pipeline import run_pipeline
from phd_agentic.fcnp import fcnp_prune

trace = run_pipeline('What is the current mandi price of wheat in Punjab?')
stats = trace['fcnp']['stats']
print(stats)
assert stats['pinned'] >= 1
assert all(item.get('pinned') for item in trace['fcnp']['retained'] if item['kind']=='citation') or True
noise = [{'id': f'n{i}', 'text': 'noise', 'pinned': False, 'feedback': -1, 'accessCount': 0, 'importance': 0, 'kind': 'consensus'} for i in range(10)]
pinned = [{'id': 'cite', 'text': 'citation:datagov.mandi_prices', 'pinned': True, 'feedback': 1, 'accessCount': 1, 'importance': 0.2, 'kind': 'citation'}]
pruned = fcnp_prune(pinned + noise, prune_ratio=0.8)
assert any(item['id']=='cite' for item in pruned['retained'])
print('FCNP OK')
```
